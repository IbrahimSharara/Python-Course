**4-Write a Python function that takes a list of words and return the longest word and the length of the longest one**

```python
def TallestWord(words):
    tallest = ""
    for word in words:
        if len(word) > len(tallest):
            tallest = word
    return tallest
wordsCount = int(input("Enter the number of words: "))
words = set()
for i in range(wordsCount):
    word = input(f"Enter word {i + 1}: ")
    words.add(word)
tallest = TallestWord(words)
print("The tallest word is:", tallest , "with length:", len(tallest))
```

**55-Write a Python program to calculate the average value of the numbers in a given tuple of tuples**

```python
myTuple =((1,2,3) , (1,1,1) , (2,2,2))
resultTuple = []
for item in myTuple:
    size = len(item)
    sum = 0
    for i in item :
        sum+= i
    average = sum / size
    resultTuple.append(average)
print(resultTuple)
```
